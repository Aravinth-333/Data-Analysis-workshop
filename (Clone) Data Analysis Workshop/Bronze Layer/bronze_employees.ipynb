### Step 1 - Define the source file

```python
source_path = "/Volumes/workspace/data_analysis/employee_dataset/Employee_Data_India (1).csv"
```

```python
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType
)

schema = StructType([
    StructField("Employee_ID", StringType(), True),
    StructField("First_Name", StringType(), True),
    StructField("Last_Name", StringType(), True),
    StructField("Email", StringType(), True),
    StructField("Phone_Number", StringType(), True),
    StructField("Department_Region", StringType(), True),
    StructField("Birth_Year", StringType(), True),
    StructField("Age", StringType(), True),
    StructField("Join_Date", StringType(), True),
    StructField("Salary", StringType(), True),
    StructField("Performance_Score", StringType(), True),
    StructField("Remote_Work", StringType(), True),
    StructField("Status", StringType(), True)
])
```

### Step 2 - Read the source CSV

```python
df_raw = (
    spark.read
    .option("header",True)
    .schema(schema)
    .csv(source_path)
)
```

### Step 3 — Inspect the Bronze input

```python
df_raw.printSchema()
```

```python
df_raw.show()
```

```python
print("Source row count:",df_raw.count())
```

### Step 4 — Add Bronze metadata

```python
from pyspark.sql.functions import current_timestamp, lit

batch_id = "BATCH_001"
source_system = "employee_app"

df_bronze = (
    df_raw
    .withColumn("batch_id", lit(batch_id))
    .withColumn("source_system", lit(source_system))
    .withColumn("ingestion_timestamp", current_timestamp())
)
```

### Step 5 — Check the Bronze Data

```python
display(df_bronze.limit(10))
```

```python
df_bronze.printSchema()
```

### Step 6 — Check Bronze row count

```python
bronze_count = df_bronze.count()
print("Bronze row count:",bronze_count)
```

```python
source_count = df_raw.count()
bronze_count = df_bronze.count()

print("Source count:",source_count)
print("Bronze count:",bronze_count)
```

### Step 7 — Write the Bronze Delta Table

```python
bronze_table = 'workspace.data_analysis.bronze_employees'

(
    df_bronze
    .write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(bronze_table)
)
```
